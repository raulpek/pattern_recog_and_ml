```python
import pandas as pd
import numpy as np
```

```python
def calculate_entropy(P1, P2):
  if P1 == 0 or P2 == 0:
        return 0
  else:
    I_N = - P1 * np.log2(P1) - P2 * np.log2(P2)
    return I_N
def calculate_info_gain(root_ent, f_weighted):
  return root_ent - f_weighted

df = pd.read_csv("t116.csv")
print(f"The number of rows: {len(df)}")
print(f"unique values per column: {df.nunique()}") # each column has two possible values
num_classes = 2
features = df.iloc[:,0:3]
labels = df.iloc[:,3]
# root entropy
p_known = (features.iloc[:,0] == 'known').sum() / len(df)
p_unknown = (features.iloc[:,0] == 'unknown').sum() / len(df)
p_new = (features.iloc[:,1] == 'new').sum() / len(df)
p_fwlp = (features.iloc[:,1] == 'follow_up').sum() / len(df)
p_short = (features.iloc[:,2] == 'short').sum() / len(df)
p_long = (features.iloc[:,2] == 'short').sum() / len(df)
# Determine the probabilities, two options must add up to 1.
# F1
p_known_reads = ((features.iloc[:,0] == 'known') & (labels == 'reads')).sum() / (features.iloc[:,0] == 'known').sum() # 0.5
p_known_skips = 1 - p_known_reads # 0.5
p_unknown_reads = ((features.iloc[:,0] == 'unknown') & (labels == 'reads')).sum() / (features.iloc[:,0] == 'unknown').sum() # 0.5
p_unknown_skips = 1 - p_unknown_reads # 0.5
# F2
p_new_reads = ((features.iloc[:,1] == 'new') & (labels == 'reads')).sum() / (features.iloc[:,1] == 'new').sum() # 0.7
p_new_skips = 1 - p_new_reads # 0.3
p_fwlp_reads = ((features.iloc[:,1] == 'follow_up') & (labels == 'reads')).sum() / (features.iloc[:,1] == 'follow_up').sum() # 0.25
p_fwlp_skips = 1 - p_fwlp_reads # 0.75
# F3
p_short_reads = ((features.iloc[:,2] == 'short') & (labels == 'reads')).sum() / (features.iloc[:,2] == 'short').sum() # 0.82
p_short_skips = 1 - p_short_reads # 0.18
p_long_reads = ((features.iloc[:,2] == 'long') & (labels == 'reads')).sum() / (features.iloc[:,2] == 'long').sum() # 0.0
p_long_skips = 1 - p_long_reads # 1.0
# calculate the entropies
entropy_known = calculate_entropy(p_known_reads, p_known_skips)
entropy_unknown = calculate_entropy(p_unknown_reads, p_unknown_skips)
entropy_new = calculate_entropy(p_new_reads, p_new_skips)
entropy_fwlp = calculate_entropy(p_fwlp_reads, p_fwlp_skips)
entropy_short = calculate_entropy(p_short_reads, p_short_skips)
entropy_long = calculate_entropy(p_long_reads, p_long_skips)
# calculate weighted average entropies
f1_weighted_avg_entropy = (p_known * entropy_known) + (p_unknown * entropy_unknown)
f2_weighted_avg_entropy = (p_new * entropy_new) + (p_fwlp * entropy_fwlp)
f3_weighted_avg_entropy = (p_short * entropy_short) + (p_long * entropy_long)
# so then the information gains are
root_entropy = calculate_entropy(p_known, p_unknown)
ig_f1 = calculate_info_gain(root_entropy, f1_weighted_avg_entropy)
ig_f2 = calculate_info_gain(root_entropy, f2_weighted_avg_entropy)
ig_f3 = calculate_info_gain(root_entropy, f3_weighted_avg_entropy)
# store into a list
print(f"The information gain for F1 is: {ig_f1}") # -0.08170416594551044
print(f"The information gain for F2 is: {ig_f2}") # 0.06812172361115687
print(f"The information gain for F3 is: {ig_f3}") # 0.5002723456084085
```
